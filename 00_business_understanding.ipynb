## Credit Risk Scorecard Project

### Problem Statement
A consumer lending bank needs to decide which loan applicants to approve.
Approving a bad applicant (who defaults) costs the bank the loan principal.
Rejecting a good applicant (who would have repaid) costs the bank lost interest income.

The goal is to build a model that predicts default probability and use it
to make approval decisions that **maximise expected profit** while managing risk.

### What is "Default"?
In this dataset, TARGET = 1 means the client had payment difficulties:
- Late payment of more than X days on at least one installment
TARGET = 0 means all payments were made on time.

### Cost Matrix
The two types of errors have very different costs:

| Decision | Reality | Outcome | Cost |
|----------|---------|---------|------|
| Approve | Repays (0) | Correct  | +$Interest earned |
| Approve | Defaults (1) | False Negative  | -$Loan principal lost |
| Reject | Would default (1) | Correct  | $0 |
| Reject | Would repay (0) | False Positive  | -$Interest foregone |

**Assumed values (illustrative):**
- Average loan: $10,000
- Average interest margin: 10% = $1,000 profit per good loan
- Loss given default: 60% = $6,000 loss per bad loan

A false negative is **6x more costly** than a false positive.
This means we should set our approval threshold conservatively.

```python
# Quantify the cost matrix
avg_loan = 10000
interest_margin = 0.10
loss_given_default = 0.60

profit_good_loan = avg_loan * interest_margin
loss_bad_loan = avg_loan * loss_given_default

print(f"Profit per good loan approved: ${profit_good_loan:,.0f}")
print(f"Loss per bad loan approved: ${loss_bad_loan:,.0f}")
print(f"Cost ratio (bad/good): {loss_bad_loan/profit_good_loan:.1f}x")
print(f"\nThis means we need {loss_bad_loan/profit_good_loan:.1f} good loans")
print(f"to offset every 1 bad loan we approve.")
```

### Success Metrics
We will NOT use accuracy (useless with 8% default rate).

Instead we evaluate using:
1. **KS Statistic** — separation between default and non-default score distributions
2. **Gini Coefficient** — overall discriminatory power (industry standard)
3. **Calibration** — do predicted probabilities match actual default rates?
4. **Expected Profit** — at our chosen threshold, what is profit per 1000 applicants?

A model is only useful if it translates into a better lending decision than
a naive "approve everyone" or "approve nobody" strategy.
