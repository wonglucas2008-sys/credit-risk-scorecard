# Data Understanding
## Home Credit Default Risk
This notebook covers CRISP-DM Phase 2 — understanding the structure, 
quality, and content of the data before any modelling.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option('display.max_columns', 100)
pd.set_option('display.float_format', '{:.2f}'.format)

print("Libraries loaded successfully")
```

```python
import os

# Check where Python thinks we are
print(os.getcwd())

# Load with relative path
df = pd.read_csv('data/home-credit-default-risk/application_train.csv')
print(f"Shape: {df.shape}")
print(f"Rows: {df.shape[0]:,} | Columns: {df.shape[1]}")
```

```python
df.head()
```

```python
# Check target variable (1 = defaulted, 0 = repaid)
print("Target distribution:")
print(df['TARGET'].value_counts())
print(f"\nDefault rate: {df['TARGET'].mean():.2%}")
```

```python
# Check missing values
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)
missing_df = pd.DataFrame({
    'missing_count': missing,
    'missing_pct': missing_pct
}).query('missing_count > 0').sort_values('missing_pct', ascending=False)

print(f"Columns with missing values: {len(missing_df)}")
missing_df.head(20)
```

```python
os.makedirs('outputs', exist_ok=True)

# Visualise target imbalance
fig, ax = plt.subplots(figsize=(6, 4))
df['TARGET'].value_counts().plot(kind='bar', ax=ax, color=['steelblue', 'crimson'])
ax.set_title('Loan Default Distribution')
ax.set_xlabel('TARGET (0=Repaid, 1=Defaulted)')
ax.set_ylabel('Count')
ax.set_xticklabels(['Repaid (0)', 'Defaulted (1)'], rotation=0)
for p in ax.patches:
    ax.annotate(f'{p.get_height():,.0f}', 
                (p.get_x() + p.get_width()/2, p.get_height()),
                ha='center', va='bottom')
plt.tight_layout()
plt.savefig('outputs/01_target_distribution.png', dpi=150)
plt.show()
```
